```
#sankey graph
import pandas as pd
import plotly.graph_objects as go
```

```
# 读取Excel文件
data = pd.read_excel("C://Users//bryennt//Desktop//employment_by_sector(unit_million people).xlsx") 
```

```
# 获取年份列和三个产业的列数据
years = data['year'].tolist()
primary = data['primary'].tolist()
secondary = data['secondary'].tolist()
tertiary = data['tertiary'].tolist()
```

```
# 创建节点标签
nodes = []
for i in range(24):  # 2000–2023 共24年
    nodes.append(f'{years[i]} - Primary')
    nodes.append(f'{years[i]} - Secondary')
    nodes.append(f'{years[i]} - Tertiary')
```

```
# 定义节点颜色
node_colors = []
sector_colors = ["#226a46", "#baad4e", "#f4e482"]  # Primary, Secondary, Tertiary
for i in range(24):
    node_colors.extend(sector_colors)
```

```
# 创建源节点、目标节点和流动的值 + hover标签
sources, targets, values, labels, link_colors = [], [], [], [], []
```

```
for i in range(23):  # 2000-2022流动
    year_idx = i * 3  # 每年3个节点
    total_next = primary[i + 1] + secondary[i + 1] + tertiary[i + 1]
    current_sectors = [primary[i], secondary[i], tertiary[i]]
    next_sectors = [primary[i + 1], secondary[i + 1], tertiary[i + 1]]
    for src_sector in range(3):
        for tgt_sector in range(3):
            src = year_idx + src_sector
            tgt = year_idx + 3 + tgt_sector
            value = next_sectors[tgt_sector] / total_next * current_sectors[src_sector]
            sources.append(src)
            targets.append(tgt)
            values.append(value)
            labels.append(
                f"{nodes[src]} → {nodes[tgt]}<br>{value:.2f} million people"
            )
            link_colors.append(sector_colors[src_sector])
```

```
# 创建图形对象
fig = go.Figure(go.Sankey(
    arrangement="snap",
    node=dict(
        pad=30,
        thickness=20,
        line=dict(color="black", width=0.5),
        label=[""] * len(nodes),
        color=node_colors
    ),
    link=dict(
        source=sources,
        target=targets,
        value=values,
        color=link_colors,
        customdata=labels,  # 传入你原来的 label
        hovertemplate="%{customdata}<extra></extra>"  # 只显示自定义 hover label
    )
))
```

```
# 添加图例（放在底部）
fig.add_annotation(
    x=0.5,
    y=-0.18,
    text=(
        "<span style='color:#226a46;'>■ Primary</span> &nbsp;&nbsp; "
        "<span style='color:#baad4e;'>■ Secondary</span> &nbsp;&nbsp; "
        "<span style='color:#f4e482;'>■ Tertiary</span>"
    ),
    showarrow=False,
    font=dict(size=14),
    align="center",
    xref="paper",
    yref="paper"
)
```

```
# 添加年份标签
for i, year in enumerate(range(2000, 2024)):
    fig.add_annotation(
        x=i / 23,
        y=-0.07,
        text=str(year),
        showarrow=False,
        font=dict(size=12),
        xref="paper",
        yref="paper"
    )
```

```
# 设置标题和布局
fig.update_layout(
    title_text="Employment Sector Transitions in China (2000–2023)",
    font_size=14,
    margin=dict(l=50, r=50, t=100, b=150),
    height=750
)
```
