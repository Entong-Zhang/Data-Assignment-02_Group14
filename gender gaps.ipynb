```
import pandas as pd
import plotly.express as px
import webbrowser
```

```
# 读取美国男性
us_men = pd.read_excel("C:\\Users\\38670\\Documents\\Documents\\Labor Force Participation Rate - Men 年度数据(LNS11300001).xlsx")
us_men = us_men.rename(columns={"observation_date": "Year", "LNS11300001": "Rate"})
us_men["Year"] = pd.to_datetime(us_men["Year"]).dt.year
us_men["Country"] = "USA"
us_men["Gender"] = "Male"
```

```
# 读取美国女性
us_women = pd.read_excel("C:\\Users\\38670\\Documents\\Documents\\Labor Force Participation Rate - Women年度 (LNS11300002).xlsx")
us_women = us_women.rename(columns={"observation_date": "Year", "LNS11300002": "Rate"})
us_women["Year"] = pd.to_datetime(us_women["Year"]).dt.year
us_women["Country"] = "USA"
us_women["Gender"] = "Female"
```

```
# 读取中国男性
cn_men = pd.read_excel("C:\\Users\\38670\\Documents\\Documents\\劳动人口参与率（15岁以上男性）.xlsx", skiprows=5)
cn_men = cn_men.rename(columns={cn_men.columns[0]: "Year", cn_men.columns[1]: "Rate"})
cn_men["Year"] = cn_men["Year"].astype(str).str.extract(r"(\d{4})")  # 提取年份中的4位数字
cn_men["Country"] = "China"
cn_men["Gender"] = "Male"
```

```
# 读取中国女性
cn_women = pd.read_excel("C:\\Users\\38670\\Documents\\Documents\\劳动人口参与率（15岁以上女性）.xlsx", skiprows=5)
cn_women = cn_women.rename(columns={cn_women.columns[0]: "Year", cn_women.columns[1]: "Rate"})
cn_women["Year"] = cn_women["Year"].astype(str).str.extract(r"(\d{4})")
cn_women["Country"] = "China"
cn_women["Gender"] = "Female"
```

```
# 合并所有数据
df = pd.concat([us_men, us_women, cn_men, cn_women], ignore_index=True)
```

```
df = df.dropna(subset=["Year", "Rate"])
df["Year"] = pd.to_numeric(df["Year"], errors="coerce")
df["Rate"] = pd.to_numeric(df["Rate"], errors="coerce")
df = df.dropna(subset=["Year", "Rate"])
df["Year"] = df["Year"].astype(int)
```

```
df = df[(df["Year"] >= 2000) & (df["Year"] <= 2023)]
```

```
fig = px.bar(
    df,
    x="Year",
    y="Rate",
    color="Gender",
    facet_col="Country",
    barmode="stack",
    title="Labor Force Participation Rate by Gender (China vs. USA)",
    labels={"Rate": "Participation Rate (%)"},
    height=600
)
```

```
output_path = "lfpr_gender_stack_plotly.html"
fig.write_html(output_path)
webbrowser.open(output_path)
```
