```
import pandas as pd
import plotly.graph_objects as go

# 工作路径
work_path = r'C:\Users\小小\PycharmProjects\第二次作业'
file_name = '美国劳动力参与率.csv'

# 读取数据
try:
    df = pd.read_csv(f'{work_path}\\{file_name}')
except FileNotFoundError:
    print(f"File {file_name} not found. Please check the file name and path.")
else:
    # 提取男性和女性劳动力参与率数据
    female_data = df[(df['Disaggregation'] == 'female, National, 15+') & (df['Country Name'] == 'United States')][
        ['Year', 'Value']].rename(columns={'Value': 'Female_Participation_Rate'})
    male_data = df[(df['Disaggregation'] == 'male, National, 15+') & (df['Country Name'] == 'United States')][
        ['Year', 'Value']].rename(columns={'Value': 'Male_Participation_Rate'})

    # 合并数据
    merged_data = pd.merge(female_data, male_data, on='Year')

    # 创建交互式堆积条形图
    fig = go.Figure()

    fig.add_trace(go.Bar(
        y=merged_data['Year'],
        x=merged_data['Female_Participation_Rate'],
        orientation='h',
        name='Female Labor Force Participation Rate'
    ))

    fig.add_trace(go.Bar(
        y=merged_data['Year'],
        x=merged_data['Male_Participation_Rate'],
        orientation='h',
        name='Male Labor Force Participation Rate',
        offsetgroup=1,
        base=merged_data['Female_Participation_Rate']
    ))

    # 更新布局
    fig.update_layout(
        barmode='stack',
        title='Gender Difference in Labor Force Participation Rate in the United States from 2000 - 2023',
        xaxis_title='Labor Force Participation Rate (%)',
        yaxis_title='Year',
        font=dict(size=10)
    )

    # 保存为 HTML 文件
    fig.write_html('labor_force_participation_interactive.html')
    print("Interactive HTML file saved as labor_force_participation_interactive.html")
```
