```
import pandas as pd
import plotly.graph_objects as go
import os


def draw_employment_heatmap():
    # 假设工作路径和文件名
    work_path = r'C:\Users\小小\PycharmProjects\第二次作业'
    file_name = '最新行业数据.xlsx'

    # 读取数据
    file_path = os.path.join(work_path, file_name)
    try:
        df = pd.read_excel(file_path)
    except FileNotFoundError:
        print(f"未找到文件 {file_name}，请检查文件名和路径。")
        return

    # 提取农业、工业和服务业数据，并将年份设置为索引
    us_sector_data = df[['Year', 'Agriculture', 'Industry', 'Services']].set_index('Year')

    # 查看数据的最小值和最大值
    min_value = us_sector_data.min().min()
    max_value = us_sector_data.max().max()

    # 创建交互式热力图
    fig = go.Figure(data=go.Heatmap(
        z=us_sector_data.values,
        x=us_sector_data.columns,
        y=us_sector_data.index,
        colorscale='RdYlBu_r',
        zmin=min_value,
        zmax=max_value,
        colorbar=dict(
            tickmode='array',
            tickvals=[min_value, max_value],
            ticktext=[f"{min_value:.2f}", f"{max_value:.2f}"],
            tickfont=dict(size=10)
        ),
        hoverongaps=False
    ))

    # 添加数据标注
    for i in range(len(us_sector_data.index)):
        for j in range(len(us_sector_data.columns)):
            value = us_sector_data.iloc[i, j]
            fig.add_trace(go.Scatter(
                x=[us_sector_data.columns[j]],
                y=[us_sector_data.index[i]],
                text=[f"{value:.2f}"],
                mode='text',
                textfont=dict(
                    size=10,
                    color='white' if (value - min_value) / (max_value - min_value) < 0.5 else 'black'
                )
            ))

    # 设置英文标题和坐标轴标签
    fig.update_layout(
        title='Employment by sectors in US(2000 - 2023)',
        xaxis_title='Industry',
        yaxis_title='Year',
        font=dict(size=10),
        width=600,
        height=400
    )

    # 保存为 HTML 文件
    fig.write_html('us_employment_heatmap_interactive.html')
    print("Interactive HTML file saved as us_employment_heatmap_interactive.html")


if __name__ == "__main__":
    draw_employment_heatmap()
```
